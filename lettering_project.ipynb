<p>The object chosen for this project is <b>GB 247 Euing 26</b>, <i>Illuminations on vellum</i>, 1300-1833. This is a manuscript extracted from a 14th Century Carmelite, showcasing large printed alphabetical letters and designs. The language is in Latin (according to the Glasgow University Archives page), which can also been seen through multiple unrecognizable letter designs in the object.</p>

<p>The Glasgow University Archives page also tells us that the author of this object is <b>Esther Cory</b>; who is said to have assembelled the collection in the early nineteenth century (Cleaver et al., 2024). This collection was then sold in 1857 after a disposal of book collector <b>Philip Hanrott's</b> library, following his death - then finally bequeathed from <b>William Euing</b> to <b>Glasgow University Library</b>, where it remains to this day (Connolly, 2022).<p>

<h4><u>"Why this object?", and the Lovelace-Boden-Ridler Test</u></h4>

<p>The objective of combining generative AI with this object is to understand deeper what context these pieces were made in, what certain symbols mean, and to determine how AI and human collaboration can result in something new, or perhaps even 'creative'. <h2>2. Methodology</h2>

<p>To reach the established objectives of this project, the human creative processes of sound, image, and text generation was an output I intended to demonstrate. To do this, the GAI (Microsoft Copilot and DALL-E 2) was asked to determine the meaning and language of latin-based letter designs. These models were also asked to evaluate the drawings inside of these characters - what they depict, what they mean, and if they can relate to any real events or depictions that we know of already.<p>

<p>To  further understand the meaning of the text in this object, the GAI(s) were then asked to create a transcription of the letters which I do not recognize, and then asked to analyse what the meanings of these are. This  was done so that, when relating this object to other works, there was a baseline of latin which was now known to be in this object.</p>

<p>Finally, these outputs will be analysed and discussed in reference to how it relates to human creativity, and how these processes augmented not only our understanding of the object, but also our view of creativity as a whole.</p>

<h4>Microsoft Copilot</h4>

<p>Microsoft Copilot, in simple terms, is a version of <b>generative AI</b>. Generative AI refers to a subset of artficial intelligence specificallt focused on the <b>creation</b> of content (Picano 2024). </p>

<p>Different versions of Microsoft Copilot use different LLMS (large language models) - such as one version using OpenAI's ChatGPT4. However, the free version of Microsoft Copilot, the one which will be used, uses DALL-E 3's image AI large langugae model, also developed by Open AI (Stratton, 2024).<p>

<p><u>But how does Microsoft Copilot (DALL-E 3) even work?</u></p>

<p><img src="DALL-E2.png" alt="DALL-E 2 diagram"></img></p>

<p><a href="https://www.assemblyai.com/blog/how-dall-e-2-actually-works#what-is-dall-e-3">Navigate to the source here!</a></p>

<p>DALL-E models are learned by an OpenAI model called CLIP - Contrastive Language-Image Pre-Training. This model is based on learning how related a given caption is to a given image. I.e., learning that an image of a dog is, in fact, a dog, by connecting it to the caption given to it. The system then cdeclops its own understanding of concepts through more images/content, or <b>training data</b>. This is also done with words, using an encoder for text captions which are then sorted on relevance.</p>

<p>As seen in this diagram, the model then puts the inputted image(s) through a model called GLIDE, which inverts this image encoding porcess to <b>decode</b> these CLIP image embeddings, for the purpose of generating a new image that mainains features of the original.</p>

<p>Stratton (2024) also shows us the reason why this model will be included in this project - unlike other OpenAI services like ChatGPT 4, this AI tool is hosted on its own Microsoft servers, meaning that its models and data are unique to Copilot only. Open AI does not have access to them. For this reason, the data which Copilot uses must be treated as oppurtunity for different and unique results.</p>
