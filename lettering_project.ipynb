<h2>1. Introduction and Context</h2>

<p><mark>Click <a href="https://www.gla.ac.uk/collections/#/details?irn=260822&catType=C">here</a> to see the object yourself on the Glasgow University Special Collections Page!</mark></p>

<p><b>Welcome to Project: MS Euing 26, Narrative Building!</b></p>

<p>The object chosen for this project is <b>GB 247 MS Euing 26</b>, <i>Illuminations on vellum</i>, 1300-1833. This is a manuscript extracted from a 14th Century Carmelite, showcasing large printed alphabetical letters and designs in both Latin and English (as shown by University of Glasgow). The focus of this project will be focused on the drawings showcased inside a page of initials designs, and interpreting a narrative unique to the original meaning.</p>

<p>The Glasgow University Archives page also tells us that the author of this object is <b>Esther Cory</b>; who is said to have assembelled the collection this object is part of in the early nineteenth century (Cleaver et al., 2024). This collection was then sold in 1857 after a disposal of book collector <b>Philip Hanrott's</b> library, following his death - then finally bequeathed from <b>William Euing</b> to the <b>Glasgow University Library</b>, where it remains to this day (Connolly, 2022).<p>

<p><u>Discussion of Literature</u></p>

<p>The Lovelace Test was originally suggested by Selmer Bringsjord (2001), where the test would only be passed "if the AI's programmer could not explain how it came up with its answer". This was a test of creativity within artificial intelligence, which was then updated to the Lovelace Test 2.0 test by Mark Reidl (2014). This new test focused on if AI would conform to a set of possible constraints wihtin a prompt, having created something that a human evaluator is satisfied with. For the sake of the project, the following human-AI collaborative constrictions have been set:</p>

<ul>

<li>Image outputs must conform to the style and established context of MS Euing 26 given the data, i.e. following assumptions of a 14th century Carmelite,</li> 

<li>Full output must include a story-narrative following several re-created initials from MS Euing 26.</li> 

</ul>

<p>Boden's (2003) view on creativity is, for the sake of this project, divided into three factors - "Creativity is the ability to come up with ideas or artefacts that are <mark>new</mark>, <mark>surprising</mark>, and <mark>valuable</mark>." These factors will also be evaluated within the collaboration of myself and AI in this project.</p>

<p>Finally, Ridler's (2021) interpretaion of artifical intelligence creativity argues that the sources of which AI trains itself on are pre-made, human sources. Because of that, the output of creative media will always be human-based - therefore biased and influenced by existing ideas. It is also argued that methods of physical creation such as using materials and the ability to change your mind have served as defining factors of human creation. This makes the question of AI-creativity unique - what creative potential lies in the collaboration of human and AI with regards to this unique technological medium?</p>

<p><u>So, what <i>is</i> the Lovelace-Boden-Ridler test?</u></p>

<p>The Lovelace-Boden-Ridler test is an evaluation of human-AI collaboration, rather than just AI. To evaluate the results of this project, we will look to Boden's (2003) criteria of creation, influenced by the constraints of the Lovelace 2.0 test, and valued through Ridler's points of a different medium and agency in both parties</p>

<p>The full objectives of this project are as follows:</p>

<ul>

<li>To understand further what certain letters and designs could mean or relate to,</li>
<li>Reimagine parts of the object through a new medium with respect to the context of the drawings used,</li>
<li>Work with AI to create a new unique narrative from the drawings showcased,</i>
<li>Create a final output result to showcase this  narrative through re-created image and story,</li>
<li>Determine whether human and AI collaboration is capable of passing the Lovelace-Boden-Ridler test, through constraints, creative evaluation, and agency.</li>

</ul>

<h2>2. Methodology</h2>

<p><u>Tools Used</u></p>

<h4>Microsoft Copilot</h4>

<p>Microsoft Copilot, in simple terms, is a version of <b>generative AI</b>. Generative AI (GAI) refers to a subset of artficial intelligence specificallt focused on the <b>creation</b> of content (Picano 2024). </p>

<p>Different versions of Microsoft Copilot use different LLMS (large language models) - such as one version using OpenAI's ChatGPT4. However, the free version of Microsoft Copilot, the one which will be used, uses DALL-E 3's image AI large langugage model, also developed by Open AI (Stratton, 2024).<p>

<p><u>But how does Microsoft Copilot (DALL-E 3) even work?</u></p>

<p><img src="DALL-E2.png" alt="DALL-E 2 diagram"></img></p>

<p><a href="https://www.assemblyai.com/blog/how-dall-e-2-actually-works#what-is-dall-e-3">Navigate to the source of this image here!</a></p>

<p>DALL-E models are learned by an OpenAI model called CLIP - Contrastive Language-Image Pre-Training. This model is based on learning how related a given caption is to a given image. I.e., learning that an image of a dog is, in fact, a dog, by connecting it to the caption given to it. The system then develops its own understanding of concepts through identifying and relating captions to more images/content, or <b>training data</b>. This is also done with words, using an encoder for text captions which are then sorted through relevance.</p>

<p>As seen in this diagram, the model then puts the inputted image(s) through a model called GLIDE, which inverts this image encoding porcess to <b>decode</b> these CLIP image embeddings, for the purpose of generating a new image that mainains features of the original. (O'Connor, 2023)</p>

<p>Stratton (2024) also shows us the reason why this model will be included in this project - unlike other OpenAI services like ChatGPT 4, this AI tool is hosted on its own Microsoft servers, meaning that its models and data are unique to Copilot only. Open AI does not have access to them. For this reason, the data which Copilot uses must be treated as oppurtunity for different and unique results.</p>

<h4>Chat GPT-4</h4>

<p><p><img src="gpt4_architecture_overview.png" alt="gpt-4 diagram"></img></p>

<p><a href="https://www.speechmatics.com/company/articles-and-news/gpt-4-how-does-it-work">Navigate to the original source here!</a>

<p>GPT-4 works similarly to Microsoft Copilot. As seen in this diagram, it takes both text and image input, and then generates a sequence of output tokens. These token embeddings (i.e. the context window, or the data which the model can "remember") are then converted into features which resemble the input sequence, and sorted into layers, each one refined using features learned from the previous layer. Finally, the features of the final layer are used to generate a sequence of output tokens. (Atkins & Hughes, 2026)</p>

<p>Copilot will be used to augment the established context found about the object MS Euing 26, then will be asked to generate a comprehensive narrative or story with analysis of the object MS Euing 26, output in text form. After which Chat-GPT and Microsoft Copilot will be implemented for image re-imagination of these drawings, with a final implementation of a narrative that connects these drawings/images together. The connection between the image re-creations and narrative analysis will be then be combined to create a presentation of a story, with both text and image. This final result will be displayed in the hope that it conforms to the objectives of the project and the passing objectives of the Lovelace-Boden-Ridler test.<p>

<p>Finally, these outputs will be analysed and discussed in reference to creative and cultural benefit, and how these processes augmented not only our understanding of the object's creative capabilities, but also our view of creativity as a whole.</p>

<p>As Ridler (2021) states, these models can often be biased due to the sources of training data still being human. This is addressed by informing the context of the object for later purposes with my own independant research, with reference to Connolly (2022) and Cleaver et al. (2024). This is done so the concern of data bias or inaccuracy is not relevant within preserving the context of the actual object.</p>

A missal is a liturgical book containing the texts for celebrating the Mass. In medieval manuscripts, the large ornamental letters—often gold, purple, or multicoloured—mark the beginnings of major sections, prayers, or chants. 

Because MS Euing 26 is a collection of initials, many pages contain single letters or monograms, not continuous prose. These letters are often framed, stylized, or intertwined with decorative motifs. They are meant to be visually striking rather than semantically dense.
