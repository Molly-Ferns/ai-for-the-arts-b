<h2>1. Context</h2>

<p><mark>Click <a href="https://www.gla.ac.uk/collections/#/details?irn=260822&catType=C">here</a> to see the object yourself!</mark></p>

<p>The object chosen for this project is <b>GB 247 Euing 26</b>, <i>Illuminations on vellum</i>, 1300-1833. This is a manuscript extracted from a 14th Century Carmelite, showcasing large printed alphabetical letters and designs. The language is in Latin (according to the Glasgow University Archives page), which can also been seen through multiple unrecognizable letter designs in the object.</p>

<p>The Glasgow University Archives page also tells us that the author of this object is <b>Esther Cory</b>; who is said to have assembelled the collection in the early nineteenth century (Cleaver et al., 2024). This collection was then sold in 1857 after a disposal of book collector <b>Philip Hanrott's</b> library, following his death - then finally bequeathed from <b>William Euing</b> to <b>Glasgow University Library</b>, where it remains to this day (Connolly, 2022).<p>

<h4><u>"Why this object?", and the Lovelace-Boden-Ridler Test</u></h4>

<p>I chose this object due to the unique creativity which it exhibited. As you can see, even one letter was given a considerable amount of time and effort into its design, despite how similar it may be from the last. Letters which we don't or do recognise alike are made with delicate designs worth understanding.</p>

<p>The objective of combining generative AI with this object is to understand deeper what context these pieces were made in, what certain symbols mean, and to determine how AI and human collaboration can result in something new, or perhaps even 'creative'. This will be done with direct reference to the <b>Lovelace-Boden-Ridler</b> test.</p>
